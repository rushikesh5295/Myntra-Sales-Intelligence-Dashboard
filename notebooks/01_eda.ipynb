```python

import pandas as pd

# Load cleaned dataset
df = pd.read_csv(
    "../data/processed/myntra_sales_cleaned.csv"
)

# 1. State-wise analysis
state_analysis = (
    df.groupby("State")
    .agg(
        Total_Orders=("Order_ID", "nunique"),
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Order_Value", "sum")
    )
    .reset_index()
    .sort_values("Total_Revenue", ascending=False)
)

print("STATE-WISE SALES ANALYSIS")
display(state_analysis.head(10))


# 2. City-wise analysis
city_analysis = (
    df.groupby(["State", "City"])
    .agg(
        Total_Orders=("Order_ID", "nunique"),
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Order_Value", "sum")
    )
    .reset_index()
    .sort_values("Total_Revenue", ascending=False)
)

print("TOP 10 CITIES")
display(city_analysis.head(10))


# 3. Category by State
category_state = (
    df.groupby(["State", "Category"])
    .agg(
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Order_Value", "sum")
    )
    .reset_index()
)

print("CATEGORY PERFORMANCE BY STATE")
display(category_state.head(15))
```

```python
print("\n========== STATE-WISE SALES ==========")
print(state_analysis.head(10).to_string(index=False))

print("\n========== TOP 10 CITIES ==========")
print(city_analysis.head(10).to_string(index=False))

print("\n========== CATEGORY BY STATE ==========")
print(category_state.head(15).to_string(index=False))
```

```python

# Select top 5 states by revenue

top_5_states = state_analysis.head(5)["State"].tolist()

# Filter category performance for these states

top_states_category = category_state[
    category_state["State"].isin(top_5_states)
]

# Display complete results

print("CATEGORY PERFORMANCE - TOP 5 STATES")

display(
    top_states_category.sort_values(
        ["State", "Total_Revenue"],
        ascending=[True, False]
    )
)
```

```python

# Identify top 5 states by revenue
top_5_states = state_analysis.head(5)["State"].tolist()

# Filter category data
top_states_category = category_state[
    category_state["State"].isin(top_5_states)
].copy()

# Sort by state and revenue
top_states_category = top_states_category.sort_values(
    ["State", "Total_Revenue"],
    ascending=[True, False]
)

# Display results
print("CATEGORY PERFORMANCE - TOP 5 STATES")

display(top_states_category)
```

```python
# Select top 5 states based on revenue
top_5_states = state_analysis.head(5)["State"].tolist()

# Filter category data for those states
top_states_category = category_state[
    category_state["State"].isin(top_5_states)
].copy()

# Display category performance
print("TOP 5 STATES:")
print(top_5_states)

print("\n========== CATEGORY PERFORMANCE - TOP 5 STATES ==========")
print(
    top_states_category
    .sort_values(["State", "Total_Revenue"], ascending=[True, False])
    .to_string(index=False)
)

# Revenue comparison table
revenue_pivot = top_states_category.pivot(
    index="Category",
    columns="State",
    values="Total_Revenue"
)

print("\n========== CATEGORY-WISE REVENUE ==========")
display(revenue_pivot)

# Quantity comparison table
quantity_pivot = top_states_category.pivot(
    index="Category",
    columns="State",
    values="Total_Quantity"
)

print("\n========== CATEGORY-WISE QUANTITY ==========")
display(quantity_pivot)
```

```python
# Create a clean category-state revenue table
revenue_pivot = top_states_category.pivot(
    index="Category",
    columns="State",
    values="Total_Revenue"
)

# Arrange categories in a consistent order
revenue_pivot = revenue_pivot.reindex(
    ["Women", "Men", "Kids", "Beauty"]
)

print("CATEGORY REVENUE BY TOP 5 STATES")
display(revenue_pivot)

# Export for future dashboard integration
revenue_pivot.to_csv(
    "../data/processed/category_state_revenue.csv"
)

print("File exported successfully!")
```
